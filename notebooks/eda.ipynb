# Healthcare AI Assistant: Exploratory Data Analysis (EDA)
## Diabetes Risk Prediction Dataset

### Objectives:
1. Inspect statistical distributions of demographic and clinical biomarkers (`age`, `bmi`, `HbA1c_level`, `blood_glucose_level`).
2. Quantify class imbalance in the target variable `diabetes`.
3. Investigate demographic distributions across genders (Male, Female, Other).
4. Audit correlation patterns and assess outliers.
5. Formulate feature engineering and preprocessing hypotheses for Phase 5.

```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is accessible
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from config.settings import RAW_DATASET_PATH, EVALUATION_DIR
sns.set_theme(style="whitegrid", palette="muted")
print("Libraries loaded. Dataset path:", RAW_DATASET_PATH)
```

### 1. Load Dataset & Basic Inspection

```python
df = pd.read_csv(RAW_DATASET_PATH)
print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns")
display(df.head())
display(df.info())
display(df.describe())
```

### 2. Target Class Imbalance Analysis

```python
counts = df["diabetes"].value_counts()
pcts = df["diabetes"].value_counts(normalize=True) * 100
print(f"Non-diabetic: {counts[0]} ({pcts[0]:.2f}%)")
print(f"Diabetic:     {counts[1]} ({pcts[1]:.2f}%)")
print(f"Imbalance Ratio: {counts[0] / counts[1]:.2f} : 1")

plt.figure(figsize=(6, 4))
sns.barplot(x=["Non-Diabetic (0)", "Diabetic (1)"], y=counts.values, palette=["#2b5c8f", "#d9534f"])
plt.title("Class Distribution (Severe Imbalance)")
plt.ylabel("Count")
plt.show()
```

### 3. Clinical Biomarkers vs Diabetes

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.histplot(data=df, x="bmi", hue="diabetes", kde=True, ax=axes[0], palette=["#2b5c8f", "#d9534f"])
axes[0].set_title("BMI Distribution")

sns.histplot(data=df, x="HbA1c_level", hue="diabetes", kde=True, ax=axes[1], palette=["#2b5c8f", "#d9534f"])
axes[1].set_title("HbA1c Level Distribution")

sns.histplot(data=df, x="blood_glucose_level", hue="diabetes", kde=True, ax=axes[2], palette=["#2b5c8f", "#d9534f"])
axes[2].set_title("Blood Glucose Level Distribution")
plt.tight_layout()
plt.show()
```

### 4. Correlation Matrix

```python
num_cols = ["age", "bmi", "HbA1c_level", "blood_glucose_level", "hypertension", "heart_disease", "diabetes"]
plt.figure(figsize=(8, 6))
sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Pearson Correlation Matrix")
plt.show()
```
